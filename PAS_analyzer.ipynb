```python
import numpy as np
import os

# ---> PASTE YOUR FOLDER PATH HERE <---
# Keep the 'r' before the quotes if you are on Windows!
data_folder = r"C:\Users\jeanv\OneDrive - Delft University of Technology\Uitwisseling - TUDelft\Courses\MEP\Experiments\PALS\PAS_060326_jvbfm"


file_mapping = {
    "VEPFIT - S316L - 3AA - 1.plo": 'AA 10 h, 2', 
    "VEPFIT - S316L - 3AA - 2.plo": 'AA 10 h, 3', 
    "VEPFIT - S316L - 3VA - 1.plo": 'VA 10 h', 
    "VEPFIT - S316L - 4VA - 1.plo": 'VA 2.5 h', 
    "VEPFIT - S316L - annealed 650C 10h - 1.plo": 'AA 10 h, 1',
    "VEPFIT-S316L - as-received polished - 1.plo": 'P-1', 
    "VEPFIT - S316L - as-received polished and washed - 1.plo": 'P-2',
    "VEPFIT - S316 - irrad - 1dpa - 2.plo": '1dpa, 2', 
    "VEPFIT - S316-irrad-1dpa - pos1 - 1.plo": 'Pos1 1dpa', 
    "VEPFIT - S316-irrad-1dpa - pos2 - 1.plo": 'Pos2 1dpa',
    "VEPFIT - S316-corroded-24h-1 v2.plo": 'C-1',
    "VEPFIT - S316-corroded-24h-2 v2.plo": 'C-2',
    "VEPFIT - S316-irrad-1dpa-corroded-24h - 1, fix L+ to value from Corr 1 and 2.plo": "IC 1dpa 24 h",
    "VEPFIT - S316-irrad-4dpa-backside-1.plo": "BsA-1 10 h, 1",
    "VEPFIT - S316-irrad-4dpa-backside-scratch-1.plo": "BsA-2 10 h, 2",
    "VEPFIT - S316 - irrad - 4dpa - 1.plo": "4dpa 10 h"
}

# 1. We completely clear the dictionary to erase the old memory
vepfit_data = {}

for filename, short_name in file_mapping.items():
    full_path = os.path.join(data_folder, filename)
    
    if os.path.exists(full_path):
        try:
            raw_data = np.genfromtxt(full_path, delimiter=',')
            
            # 2. Extracting with the perfect column indices
            vepfit_data[short_name] = {
                'Energy': raw_data[:, 0], # Column 1
                'S_fit':  raw_data[:, 1], # Column 2
                'S':      raw_data[:, 2], # Column 3 (Real S)
                'W_fit':  raw_data[:, 3], # Column 4
                'W':      raw_data[:, 4], # Column 5 (Real W)
            }
        except Exception as e:
            print(f"Error processing {short_name}: {e}")

# 3. Let's print the real W array to prove it worked!
print("If this worked, these numbers should be around 0.015:")
print(vepfit_data["Pos1 1dpa"]["W"][:5]) # Prints just the first 5 W values
```

# S-W plot

Comparisons: 
1) Pristine and Annealing 
2) Annealing and Corrosion 
3) Annealing and He implantation
4) Annealing and Corrosion
5) Annealing, Implantation and Corrosion

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

class PASAnalyzer:
    def __init__(self, data_dict, rename_map, surf_e, bulk_e):
        self.data = data_dict
        self.rename_map = rename_map
        self.surf_e = surf_e
        self.bulk_e = bulk_e
        self.ref_s = 1.0
        self.ref_w = 1.0
        
        # Color logic based on your SHORT labels
        self.condition_colors = {
            'P-1': '#2ca02c',      # Pristine (Green)
            'P-2': '#2ca97c',      # Pristine (Green)
            'AA': '#1f77b4',      # Air Annealed (Blue)
            'VA': '#7497bd',      # Vacuum Annealed (Purple)
            '1dpa': '#ff7f0e',      # Irradiated (Orange)
            'Pos1': '#ffbb78',  # Irradiated Pos1 (Light Orange)
            'Pos2': '#ff9866',  # Irradiated Pos2 (Lighter Orange)
            'BsA-1': "#781fb4",      # Backside Annealed (Dark Purple)
            'BsA-2': "#ad5ee1",      # Backside Annealed (Light Purple)
            '4dpa': "#E9FF6D",     # 4dpa (Medium Purple)
            'C-1': '#d62728',      # Corroded (Red)
            'C-2': "#ff2a26",     # Corroded 2 (Light Red)
            'IC': '#8c564b',      # Irradiated + Corroded (Brown)
        }

    def set_reference(self, sample_key):
        df = self.data[sample_key]
        idx = (df['Energy'] - self.bulk_e).abs().idxmin()
        self.ref_s = df.iloc[idx]['S']
        self.ref_w = df.iloc[idx]['W']
        print(f"Reference set to: {self.rename_map.get(sample_key)}")

    def _get_color(self, label):
        for key, color in self.condition_colors.items():
            if label.startswith(key):
                return color
        return '#7f7f7f' 

    def plot_comparison(self, keys_to_plot, title="S-W Parameter Map"):
        plt.figure(figsize=(10, 9))
        
        for key in keys_to_plot:
            if key not in self.data: continue
            
            label = self.rename_map.get(key, key)
            df = self.data[key]
            color = self._get_color(label)
            
            # Indices for Surface and Bulk
            idx_s = (df['Energy'] - self.surf_e).abs().idxmin()
            idx_b = (df['Energy'] - self.bulk_e).abs().idxmin()
            
            # 1. Extract Values and Errors
            s_raw = np.array([df.iloc[idx_s]['S'], df.iloc[idx_b]['S']])
            w_raw = np.array([df.iloc[idx_s]['W'], df.iloc[idx_b]['W']])
            s_err_raw = np.array([df.iloc[idx_s]['S_error'], df.iloc[idx_b]['S_error']])
            w_err_raw = np.array([df.iloc[idx_s]['W_error'], df.iloc[idx_b]['W_error']])

            # 2. Normalize Everything
            s_vec = s_raw / self.ref_s
            w_vec = w_raw / self.ref_w
            s_err = s_err_raw / self.ref_s
            w_err = w_err_raw / self.ref_w

            s_vec = s_raw 
            w_vec = w_raw 
            s_err = s_err_raw 
            w_err = w_err_raw
            
            # 3. Plot with Error Bars
            # Plot Bulk (Circle) - Index 1
            plt.errorbar(w_vec[1], s_vec[1], xerr=w_err[1], yerr=s_err[1], 
                            fmt='o', color=color, ms=10, mec='k', capsize=3, label=f"{label} bulk")
                
                # Plot Surface (Diamond) - Index 0
            plt.errorbar(w_vec[0], s_vec[0], xerr=w_err[0], yerr=s_err[0], 
                            fmt='D', color=color, ms=8, alpha=0.6, mec='k', capsize=3, label=f"{label} surface")
                
                # Draw the connecting line
            plt.plot(w_vec, s_vec, color=color, lw=1.5, alpha=0.3)


        # Reference Lines at (1,1)
        plt.axhline(1, color='k', lw=0.5, ls='--')
        plt.axvline(1, color='k', lw=0.5, ls='--')
        
        plt.xlabel("W", fontsize=12)
        plt.ylabel("S", fontsize=12)
        plt.xlim(0.011,0.040)
        plt.ylim(0.49,0.60)
        plt.title(title, fontsize=14)
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Samples")
        plt.grid(True, ls=':', alpha=0.4)
        plt.tight_layout()
        plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

class PASAnalyzer:
    def __init__(self, data_dict, rename_map, surf_e, bulk_e, surf_e_min, surf_e_max, bulk_e_min, bulk_e_max):
        self.data = data_dict
        self.rename_map = rename_map
        self.surf_e = surf_e
        self.bulk_e = bulk_e
        self.surf_e_min = surf_e_min
        self.surf_e_max = surf_e_max
        self.bulk_e_min = bulk_e_min
        self.bulk_e_max = bulk_e_max
        self.ref_s = 1.0
        self.ref_w = 1.0
        
        # Color logic based on your SHORT labels
        self.condition_colors = {
            'P-1': '#2ca02c',      # Pristine (Green)
            'P-2': '#2ca97c',      # Pristine (Green)
            'AA': '#1f77b4',      # Air Annealed (Blue)
            'VA': '#7497bd',      # Vacuum Annealed (Purple)
            '1dpa': '#ff7f0e',      # Irradiated (Orange)
            'Pos1': '#ffbb78',  # Irradiated Pos1 (Light Orange)
            'Pos2': '#ff9866',  # Irradiated Pos2 (Lighter Orange)
            'BsA-1': "#781fb4",      # Backside Annealed (Dark Purple)
            'BsA-2': "#ad5ee1",      # Backside Annealed (Light Purple)
            '4dpa': "#E9FF6D",     # 4dpa ()
            'C-1': '#d62728',      # Corroded (Red)
            'C-2': "#ff2a26",     # Corroded 2 (Light Red)
            'IC': '#8c564b',      # Irradiated + Corroded (Brown)
        }

    def set_reference(self, sample_key):
        df = self.data[sample_key]
        idx = (df['Energy'] - self.bulk_e).abs().idxmin()
        self.ref_s = df.iloc[idx]['S']
        self.ref_w = df.iloc[idx]['W']
        print(f"Reference set to: {self.rename_map.get(sample_key)}")

    def _get_color(self, label):
        for key, color in self.condition_colors.items():
            if label.startswith(key):
                return color
        return '#7f7f7f' 

# Change the def line to set range=None
    def plot_comparison(self, keys_to_plot, title="S-W Parameter Map", energy_range=None, areas=False):
        # If no range is provided, use the class defaults inside the function
        if energy_range is None:
            energy_range = [self.surf_e_min, self.surf_e_max, self.bulk_e_min, self.bulk_e_max]
            
        plt.figure(figsize=(15, 10))
        ax = plt.gca() # Get current axis for the Ellipses
        
        for key in keys_to_plot:
            if key not in self.data: continue
            
            label = self.rename_map.get(key, key)
            df = self.data[key]
            color = self._get_color(label)
            
            if areas:
                # Find indices for the ranges
                idx_s_min = (df['Energy'] - energy_range[0]).abs().idxmin()
                idx_s_max = (df['Energy'] - energy_range[1]).abs().idxmin()
                idx_b_min = (df['Energy'] - energy_range[2]).abs().idxmin()
                idx_b_max = (df['Energy'] - energy_range[3]).abs().idxmin()
              
                # Calculate means for Surface (Index 0) and Bulk (Index 1)
                surf_slice = df.iloc[idx_s_min : idx_s_max + 1]
                bulk_slice = df.iloc[idx_b_min : idx_b_max + 1]

                s_vec = np.array([surf_slice['S'].mean(), bulk_slice['S'].mean()])
                w_vec = np.array([surf_slice['W'].mean(), bulk_slice['W'].mean()])
                
                # --- NEW TOTAL UNCERTAINTY AREA LOGIC ---
                # 1. Spread of the points (Standard Deviation). Handle edge case if slice is 1 point.
                s_std_surf = surf_slice['S'].std(ddof=1) if len(surf_slice) > 1 else 0
                s_std_bulk = bulk_slice['S'].std(ddof=1) if len(bulk_slice) > 1 else 0
                w_std_surf = surf_slice['W'].std(ddof=1) if len(surf_slice) > 1 else 0
                w_std_bulk = bulk_slice['W'].std(ddof=1) if len(bulk_slice) > 1 else 0

                # 2. Average instrumental error of the points
                s_instr_surf = surf_slice['S_error'].mean()
                s_instr_bulk = bulk_slice['S_error'].mean()
                w_instr_surf = surf_slice['W_error'].mean()
                w_instr_bulk = bulk_slice['W_error'].mean()

                # 3. Combine them in quadrature (Sum of Squares)
                s_err = np.array([
                    np.sqrt(s_std_surf**2 + s_instr_surf**2),
                    np.sqrt(s_std_bulk**2 + s_instr_bulk**2)
                ])
                w_err = np.array([
                    np.sqrt(w_std_surf**2 + w_instr_surf**2),
                    np.sqrt(w_std_bulk**2 + w_instr_bulk**2)
                ])
                # ----------------------------------------
                
            else:
                # Original Point-Lookup Logic
                idx_s = (df['Energy'] - self.surf_e).abs().idxmin()
                idx_b = (df['Energy'] - self.bulk_e).abs().idxmin()
                s_vec = np.array([df.iloc[idx_s]['S'], df.iloc[idx_b]['S']])
                w_vec = np.array([df.iloc[idx_s]['W'], df.iloc[idx_b]['W']])
                s_err = np.array([df.iloc[idx_s]['S_error'], df.iloc[idx_b]['S_error']])
                w_err = np.array([df.iloc[idx_s]['W_error'], df.iloc[idx_b]['W_error']])


            # --- PLOTTING ---
            if not areas:
                plt.errorbar(w_vec[1], s_vec[1], xerr=w_err[1], yerr=s_err[1], 
                            fmt='o', color=color, ms=10, mec='k', capsize=3, label=f"{label} bulk")
                plt.errorbar(w_vec[0], s_vec[0], xerr=w_err[0], yerr=s_err[0], 
                            fmt='D', color=color, ms=8, alpha=0.6, mec='k', capsize=3, label=f"{label} surface")
                plt.plot(w_vec, s_vec, color=color, lw=1.5, alpha=0.3)

            else:
                # Plot central points
                plt.plot(w_vec[1], s_vec[1], 'o', ms=10, mec='k', color=color, label=f"{label} bulk area")
                plt.plot(w_vec[0], s_vec[0], 'D', ms=10, alpha=0.6, mec='k', color=color, label=f"{label} surface area")

                # Add Ellipses (using w_err for width and s_err for height)
                for i in [0, 1]:
                    # Width/Height of Ellipse usually represent 2*sigma for the full diameter
                    ell = Ellipse(xy=(w_vec[i], s_vec[i]), 
                                  width=w_err[i]*2, height=s_err[i]*2,
                                  angle=0, edgecolor=color, facecolor=color, 
                                  alpha=0.2, linestyle='--')
                    ax.add_patch(ell)

        # Plot Styling
        plt.xlabel("W parameter", fontsize=12)
        plt.ylabel("S parameter", fontsize=12)
        plt.xlim(0.011, 0.040)
        plt.ylim(0.49, 0.61)
        plt.title(title, fontsize=14)
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Samples")
        plt.grid(True, ls=':', alpha=0.4)
        plt.tight_layout()
        plt.show()
```
