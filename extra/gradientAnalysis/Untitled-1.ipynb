```python
import numpy as np
import matplotlib.pyplot as plt

massinitial = np.array([0.38367, 0.40468, 0.71619])
massloss = np.array([-0.00074, -0.00098, -0.0016])
massfinal = np.array([0.38293, 0.4037, 0.71459])
surface_corrosion = np.array([160, 160, 240])
surface_sample = np.array([160, 160, 240])

masslosspersurf = massloss / surface_corrosion
massinitialpersurf = massinitial / surface_sample

```

```python

plt.figure(figsize=(8, 5))
for i in range(len(massinitialpersurf)):
    plt.plot(massinitialpersurf[i], masslosspersurf[i], 'o', color=['red', 'red', 'brown'][i], label=["C-1", "C-2", "IC"][i])
plt.xlabel('Initial Mass per surface area (g/mm²)', fontsize=12)
plt.ylabel('Mass Loss per surface area (g/mm²)', fontsize=12)
plt.title('Surface Normalized Mass Loss vs. Initial Mass', fontsize=14)
plt.legend()
plt.tight_layout()
plt.show()
```
