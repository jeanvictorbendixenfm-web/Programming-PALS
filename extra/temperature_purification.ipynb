```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

```python


temperatures = [250, 300, 350, 440, 410, 425, 500, 550, 640, 600, 615, 612, 610, 608, 780, 723, 723]
times = [1170, 8750, 11600, 14320, 15450, 16560, 76320, 76810, 80320, 81380, 82020,82850,83030,91690,104000,162000, 300000]
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

# Define the absolute color scale based on temperature
cmap = plt.cm.coolwarm
norm = mcolors.Normalize(vmin=min(temperatures), vmax=max(temperatures))

# Shading each interval based on the temperature value of that step
for i in range(len(times) - 1):
    # This ensures color is tied to the temperature value, not the time index
    current_temp = temperatures[i]
    color = cmap(norm(current_temp))
    ax.axvspan(times[i], times[i+1], color=color, alpha=0.6)

# Overlay the staircase profile
ax.step(times, temperatures, where='post', color='black', linewidth=1.5, marker='.')

# Add a colorbar to show the Temperature-to-Color relationship
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label(r'Temperature ($\degree$C)')

ax.set_xlabel('Time (s)')
ax.set_ylabel(r'Temperature ($\degree$C)')
ax.set_title(r'Furnace Heating Profile $T_f$')

plt.tight_layout()
plt.show()
```
