```python
import pandas as pd
from pathlib import Path

# 1. Define your path
path_string = r"C:\Users\jeanv\OneDrive - Delft University of Technology\Uitwisseling - TUDelft\Courses\MEP\Experiments\PALS\PAS_060326_jvbfm"

# 2. CONVERT it to a Path object (This is the crucial step!)
folder_path = Path(path_string)

# 2. Setup a dictionary to hold the results
all_data = {}

# 3. Loop through every .txt file in the folder
for file in folder_path.glob("*.txt"):
    # Get the filename without the '.txt' extension to use as a key
    sample_name = file.stem 
    
    # Read the file
    df = pd.read_csv(
        file, 
        sep='\t', 
        skiprows=1, 
        header=None, 
        names=['Energy', 'S', 'S_error', 'W', 'W_error', '3g2g', '3g2g_error']
    )
    
    # Store the entire DataFrame in our dictionary
    all_data[sample_name] = df
    
    print(f"Loaded: {sample_name}")

# --- HOW TO ACCESS YOUR DATA ---
# Example: Get the 'S' column for a specific file
# If your file was "Jean_Victor.txt", the key is "Jean_Victor"
specific_s_data = all_data['Jean Victor - S316 - as-received polished']['W']
print(specific_s_data.head())

# You can even do math instantly:
# df['Time_ns'] = df['Channel'] * 0.0125  # Example conversion
```
