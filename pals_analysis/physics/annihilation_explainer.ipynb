# Annihilation Calculation




In this function, we solve the annihilation-diffusion equation for Positrons. 
This is done with the following equation; 

$$
D_+ \frac{d^2 C(z)}{dz^2} - \lambda_{eff} C(z) = -I(z)
$$
Where $D_+$ is the diffusion coefficient, $C(z)$ is the concentration profile, $\lambda_{eff}$ is the effective annihilation rate and $I(z)$ is a source term, also known as the implantation profile - essentially the Makhov profile. 

We can divide the whole equation through with $\lambda_{eff}$, and by the relation; $L_+ = \sqrt{D_+/\lambda_{eff}}$, we can write; 

$$
L_+^2 \frac{d^2 C(z)}{dz^2} - C(z) = -p(z)
$$


Using the Finite Difference Scheme and writing $C$ as $C_i$ and $P_i$
$$
\frac{d^2 C}{dz^2} \approx \frac{C_{i+1} - 2C_i + C_{i-1}}{\Delta z^2}
$$
we plug this into the equation above and assemble all the terms; 
$$
L_{+,i}^2 \left(\frac{C_{i+1} - 2C_i + C_{i-1}}{\Delta z^2}\right) - C_i = -P_i
$$
We expand and arrange all the C terms individually; 
$$
\frac{L_{+,i}^2}{\Delta z^2}C_{i+1} - \frac{2 L_{+,i}^2}{\Delta z^2}C_i + \frac{L_{+,i}^2}{\Delta z^2} C_{i-1} - C_i\\
= \frac{L_{+,i}^2}{\Delta z^2}C_{i+1} + \left(- 1 - \frac{2 L_{+,i}^2}{\Delta z^2}\right) C_i + \frac{L_{+,i}^2}{\Delta z^2} C_{i-1} =-P_i
$$

In the code, this coefficient $ \frac{L_{+,i}^2}{\Delta z^2}$ is called 'factor', but we can for simplicity call it $\alpha_i$ here. We now have the equations and we can construct with linear algebra the problem we'd like to solve, namely the concentration profile. To set up the matrix equation we are solving, we further write $(-1-2\alpha_i)$ as $\beta_i$. 

If we write out for the indeces, we get the following equations that we can write in matrix form;but the first equation involves $C_{-1}$, which is outside the grid. This point represents a vacuum point, and thus we make it into a boundary condition. We assume that any positron diffusing out of the surface is immediately captured at the surface and annihilates there. 
$$
\alpha_0 C_{-1} + \beta_0 C_{0} + \alpha_0 C_{1} = - P_{0}, i=0, C_{-1} = 0 \\
\alpha_1 C_{0} + \beta_1 C_{1} + \alpha_1 C_{2} = - P_{1}, i=1 \\
\alpha_2 C_{1} + \beta_2 C_{2} + \alpha_2 C_{3} = - P_{2}, i=2 \\
$$
We write it on matrix form; 

$$\begin{bmatrix}
\mathbf{\beta_0} & \mathbf{\alpha_0} & 0 & ... & ... &0 \\
\alpha_1 & \beta_1 & \alpha_1 & 0 & ... & 0 \\
0 & \alpha_2 & \beta_2 & \alpha_2 & ... & 0 \\
... & 0 & \alpha_3 & \beta_3 & \alpha_3 & 0 \\
... & ... & ... & ... & ... & ... \\
0 & ... & ... & 0 & \alpha_n & \beta_n
\end{bmatrix}
\cdot
\begin{bmatrix}
C_0 \\
C_1 \\
C_2 \\
C_3 \\
... \\
C_n
\end{bmatrix}$$

We then solve this matrix equation using spsolve() and then both filter out negative values (corresponding to noise and nonphysical values) and normalize it.
